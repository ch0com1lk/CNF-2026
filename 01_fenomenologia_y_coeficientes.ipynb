# ***01 · Fenomenología y coeficientes $A(\sqrt{s})$, $B(\sqrt{s})$***

**Proceso:** $e^+e^- \to \mu^+\mu^-$ · **Objetivo:** puntos 1–3 del plan (fenomenología, derivación desde la amplitud e interferencia $\gamma$–$Z$) y cálculo de los coeficientes con el propagador de Breit-Wigner.

## ***Teoría***

### ***1. Fenomenología del proceso***

A nivel árbol, el $e^+e^-$ se aniquila en un bosón virtual ($\gamma$ o $Z$) que después se convierte en un par $\mu^+\mu^-$. Se define $x=\cos\theta$, donde $\theta$ es el ángulo entre el $\mu^-$ saliente y el haz de $e^-$.

- **$x\to 1$** ($\theta\to0$): el muón sale hacia adelante (*forward*)
- **$x\to -1$** ($\theta\to\pi$): el muón sale hacia atrás (*backward*)

Con solo el fotón la distribución es simétrica. Con la $Z$ aparece un término lineal en $x$, y de ahí surge la Forward-Backward asymmetry.

### ***2. Sección eficaz a partir de la amplitud***

Se trabaja con fermiones sin masa ($m_e, m_\mu \ll \sqrt{s}$), de modo que la helicidad (*helicity*) se conserva en cada vértice. Para un electrón de quiralidad $i$ y un muón de quiralidad $j$ ($i,j=L,R$):

$$\mathcal{M}_{ij}\;\propto\;\frac{e^2}{s}\Big[\,Q_eQ_\mu + g_i^{e}\,g_j^{\mu}\,\chi(s)\Big]\times(\text{helicity spinors})$$

con el **propagador de Breit-Wigner**

$$\chi(s)=\frac{1}{\sin^2\theta_W\cos^2\theta_W}\;\frac{s}{s-m_Z^2+i\,m_Z\Gamma_Z}$$

y acoplamientos $g_L=T_3-Q\sin^2\theta_W$, $g_R=-Q\sin^2\theta_W$ ($Q=-1$, $T_3=-\tfrac12$).

El factor de los *helicity spinors*, al cuadrado, vale $(1+x)^2$ si $i=j$ ($LL,RR$) y $(1-x)^2$ si $i\neq j$ ($LR,RL$). Sumando sobre helicidades:

$$\frac{d\sigma}{dx}=\frac{\pi\alpha^2}{8s}\Big[S_{=}\,(1+x)^2+S_{\neq}\,(1-x)^2\Big],\qquad a_{ij}=Q_eQ_\mu+g_ig_j\chi(s)$$

con $S_{=}=|a_{LL}|^2+|a_{RR}|^2$ y $S_{\neq}=|a_{LR}|^2+|a_{RL}|^2$. Como $(1\pm x)^2=1+x^2\pm2x$:

$$\boxed{\frac{d\sigma}{dx}=\frac{\pi\alpha^2}{2s}\Big[A\,(1+x^2)+B\,x\Big]}\qquad A=\frac{S_{=}+S_{\neq}}{4},\quad B=\frac{S_{=}-S_{\neq}}{2}.$$

### ***3. Términos de interferencia $\gamma$–$Z$***

$$|a_{ij}|^2=\underbrace{Q_e^2Q_\mu^2}_{\gamma\gamma}\;+\;\underbrace{2\,Q_eQ_\mu\,g_ig_j\,\mathrm{Re}\,\chi}_{\text{interferencia }\gamma Z}\;+\;\underbrace{(g_ig_j)^2|\chi|^2}_{ZZ}$$

La interferencia es proporcional a $\mathrm{Re}\,\chi$, que cambia de signo al cruzar $m_Z$ y se anula en el polo. De la forma $f(x)=A(1+x^2)+Bx$ se obtiene

$$A_{FB}=\frac{N_F-N_B}{N_F+N_B}=\frac{3}{8}\frac{B}{A}.$$

## ***Implementación***

La física está en `src/fbasym/physics.py`. La función `coeficientes(raiz_s)` devuelve $A$ y $B$ para cualquier energía, con $\sqrt{s}=m_Z$ exacto en el polo, y `terminos_A(raiz_s)` separa $A$ en sus tres contribuciones.

```python
import sys
sys.path.append("../src")          # permite importar el paquete fbasym sin instalarlo

import numpy as np
import matplotlib.pyplot as plt
from fbasym import *
from fbasym import plots
```

```python
A_Z, B_Z = coeficientes(M_Z)
print(f"en el polo de la Z (raiz_s = {M_Z} GeV):")
print(f"A = {A_Z:.4f}")
print(f"B = {B_Z:.4f}")
print(f"A_FB teórica = {afb_teorica(M_Z):.4f}")

# chequeo: gamma-gamma + interferencia + ZZ = A
assert abs(sum(terminos_A(95.0)) - coeficientes(95.0)[0]) < 1e-9
```

Verificación: en el polo $A_{FB}\approx0.017$, pequeña y positiva, porque $g_V=T_3-2Q\sin^2\theta_W\approx-0.04$ es casi nulo para leptones y $A_{FB}\propto g_V^2$.

### ***Comportamiento con la energía***

```python
ECM_vals = np.linspace(40, 140, 1000)
fig = plots.figura_coeficientes(ECM_vals)
fig.savefig("../figures/01_coeficientes_vs_energia.png", dpi=150)
plt.show()
```

Observaciones:

- $A_{NC}$ presenta el pico en $m_Z$ (casi 170 veces el valor del fotón solo), mientras que $|B_{NC}|$ en el polo es mucho menor; la razón $B_{NC}/A_{NC}$ es pequeña y por eso la asimetría es chica
- la Forward-Backward asymmetry es negativa debajo de la resonancia, pasa por casi cero en el polo y crece hacia valores positivos grandes arriba de ella
- el término $\gamma\gamma$ domina lejos de la resonancia, el término $ZZ$ domina en torno a $m_Z$, y la interferencia es negativa por debajo del polo, positiva por encima y se anula en $m_Z$
