# ***03 · Transición simétrica → asimétrica***

**Objetivo:** punto 6 del plan. Se generan 10 000 eventos a varias energías, desde muy por debajo de la resonancia hasta muy por encima, para observar cómo aparece la Forward-Backward asymmetry.

```python
import sys
sys.path.append("../src")          # permite importar el paquete fbasym sin instalarlo

import numpy as np
import matplotlib.pyplot as plt
from fbasym import *
from fbasym import plots

n_events = 10000
np.random.seed(42)
```

```python
energias_demo = [30, 80, 91.1876, 95, 105, 200]
lista = []
for E in energias_demo:
    A, B = coeficientes(E)
    ev, _ = rejection_sampling(A, B, n_events)
    lista.append((E, ev))
    print(f"raiz_s = {E:8.4f} GeV   A_FB simulada = {asimetria_fb(ev):+.4f}   teórica = {afb_teorica(E):+.4f}")

fig = plots.figura_histogramas_energia(lista)
fig.savefig("../figures/03_histogramas_por_energia.png", dpi=150)
plt.show()
```

Observaciones:

- a $\sqrt{s}=30$ GeV domina el fotón: la distribución es casi la simétrica y $A_{FB}$ es chica y negativa
- cerca de la resonancia ($80$–$95$ GeV) la interferencia $\gamma$–$Z$ se intensifica y la curva se inclina: primero hacia *backward* (debajo del polo) y luego hacia *forward* (arriba)
- en $m_Z$ la interferencia es nula ($\mathrm{Re}\,\chi=0$) y la asimetría remanente proviene del término $ZZ$
- por encima de la resonancia la distribución es claramente asimétrica hacia $x>0$, con $A_{FB}\approx0.7$ a 105 GeV

### ***Barrido en energía***

Forma teórica de $f(x)$ a varias energías (izquierda) y $A_{FB}$ teórica contra simulaciones (derecha), con 10 000 eventos por punto y barra de error $\sqrt{(1-A_{FB}^2)/N}$.

```python
energias_scan = np.array([40, 60, 75, 85, 88, 90, 91.1876, 92.5, 94, 97, 105, 120, 150])
afb_mc, err_mc = [], []
for E in energias_scan:
    A, B = coeficientes(E)
    ev, _ = rejection_sampling(A, B, n_events)
    afb = asimetria_fb(ev)
    afb_mc.append(afb)
    err_mc.append(error_afb(afb, n_events))

fig = plots.figura_transicion([30, 80, M_Z, 95, 105], np.linspace(40, 140, 1000),
                              energias_scan, afb_mc, err_mc, n_events)
fig.savefig("../figures/03_transicion_y_barrido.png", dpi=150)
plt.show()
```
